A language model does not read letters or whole words. It reads tokens: small pieces of text that each have a number.

Common words are often one token. Rare words, names and many non-English scripts break into several. More tokens means more cost and less room in the context window.

**How to use:** run the cells from top to bottom (Runtime > Run all). No API key is needed.

## Lab: See the pieces

````python
import tiktoken
enc = tiktoken.get_encoding("cl100k_base")
text = "Tokens are the pieces a model reads."
ids = enc.encode(text)
print(ids)
print([enc.decode([i]) for i in ids])
for word in ["chai", "unbelievably", "नमस्ते"]:
    print(word, "->", len(enc.encode(word)), "token(s)")
````

### Expected output (from the author's run; version numbers and timings may differ)

```
[30400, 527, 279, 9863, 264, 1646, 16181, 13]
['Tokens', ' are', ' the', ' pieces', ' a', ' model', ' reads', '.']
chai -> 1 token(s)
unbelievably -> 3 token(s)
नमस्ते -> 6 token(s)
```

### Read it like this

- encode turns text into a list of numbers; decode turns one number back into its piece.
- The pieces include the space before a word, which is part of the token.
- A short Hindi word can need several tokens, so the same idea may cost more in some languages.
