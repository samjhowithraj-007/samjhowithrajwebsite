An embedding is a list of numbers that stands for the meaning of a piece of text. Texts with similar meaning get similar lists.

To compare two embeddings we measure the angle between them. With normalized vectors that is just a dot product: close to 1 means very similar.

## Lab: Which sentences mean the same?

````python
import numpy as np
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2")
sents = ["How do I reset my password?",
         "I forgot my login password",
         "Best places to eat in Pune"]
vecs = model.encode(sents, normalize_embeddings=True)
print(vecs.shape)
print(np.round(vecs @ vecs.T, 2))
````

### Expected output (from the author's run; version numbers and timings may differ)

```
(3, 384)
[[ 1.    0.78  0.03]
 [ 0.78  1.   -0.02]
 [ 0.03 -0.02  1.  ]]
```

### Read it like this

- Each sentence became a vector of 384 numbers.
- vecs @ vecs.T compares every sentence with every other one.
- The first two sentences are close even though they share few words; the food sentence is far from both.
