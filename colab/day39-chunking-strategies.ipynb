A model and a search work best on pieces of the right size: small enough to be about one thing, big enough to make sense alone.

You can cut by a fixed number of characters (with overlap) or along natural borders such as paragraphs and sentences.

**How to use:** run the cells from top to bottom (Runtime > Run all). No API key is needed.

````python
# Setup: writes the small data files this lab reads into the Colab folder
import base64, pathlib
files = {
    'policy.txt': 'Data privacy policy\n\nWe keep customer phone numbers only to send order updates. Numbers are deleted after 90 days.\n\nStaff photos are used only inside the shop. We never post a staff photo online without written consent.\n\nSurveillance cameras cover the counter and the entrance. They do not cover the staff rest room.\n\nRefund policy\n\nA drink can be replaced or refunded if the customer says it is wrong within 15 minutes of the order. Food refunds need the receipt.\n\nGift cards cannot be exchanged for cash. They are valid for one year from the date of purchase.\n\nComplaints\n\nWritten complaints go to the owner within two working days. The owner replies within seven days.\n\nAnonymous complaints are accepted through the box near the exit and are read every Friday.\n',
}
for name, content in files.items():
    p = pathlib.Path(name)
    p.write_bytes(content if isinstance(content, bytes) else content.encode('utf-8'))
print('wrote', ', '.join(files))
````

## Lab: Cut the same text two ways

````python
text = open("policy.txt", encoding="utf-8").read()

def fixed(text, size=200, overlap=40):
    step = size - overlap
    return [text[i:i + size] for i in range(0, len(text), step)]

paras = [p.strip() for p in text.split("\n\n") if p.strip()]
fx = fixed(text)
print("fixed chunks:", len(fx), "| paragraph chunks:", len(paras))
print("fixed chunk ends :", repr(fx[0][-30:]))
print("paragraph chunk  :", repr(paras[1][:60]))
````

### Expected output (from the author's run; version numbers and timings may differ)

```
fixed chunks: 5 | paragraph chunks: 10
fixed chunk ends : 'ost a staff photo online witho'
paragraph chunk  : 'We keep customer phone numbers only to send order updates. N'
```

### Read it like this

- fixed slides a 200-character window along the text, 160 characters at a time.
- The paragraph version splits on blank lines.
- The fixed chunk stops wherever the 200th character falls, here inside a word; the paragraph chunk is a whole thought.
